# Local Ollama chat

Talks to Ollama running on this machine (`http://127.0.0.1:11434`).

Start the server if it is not already running:

```bash
ollama serve
```

````python
from ollama import Client

client = Client(host="http://127.0.0.1:11434")
models = client.list()

print("Available models:")
for model in models.models:
    print(f"- {model.model}")

MODEL = models.models[0].model if models.models else None
print(f"\nUsing: {MODEL}")
````

````python
response = client.chat(
    model=MODEL,
    messages=[
        {"role": "user", "content": "In one sentence, what is a neural network?"}
    ],
)

print(response.message.content)
````

````python
print("Streaming:\n")
for chunk in client.chat(
    model=MODEL,
    messages=[{"role": "user", "content": "Name three uses of large language models."}],
    stream=True,
):
    print(chunk.message.content, end="", flush=True)
print()
````
